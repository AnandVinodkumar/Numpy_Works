```python
import pandas
from xgboost import XGBClassifier
```

```python
data = {
    "hours" : [1,2,4,5],
    "result" : [0,1,1,1]  # 0 -> Fail  1 -> Pass
}

df = pandas.DataFrame(data)

X = df["hours"]
Y = df["result"]

print(X)
print(Y)

model = XGBClassifier()
model.fit(X,Y)

Y_pred = model.predict(X)

Y_pred
```
