```python
data = pandas.read_csv('diabetes.csv')
df = pandas.DataFrame(data)
df
```

```python
# 768 rows and 9 columns
# 0  500    1  500   Over-sampling
# 1  268    0  268   Under-sampling
```

```python
X = df.iloc[:,:-1].values
Y = df.iloc[:,-1].values
print(X)
print(Y)
```

```python
from imblearn.over_sampling import SMOTE

sm = SMOTE(random_state=42)

x_new,y_new = sm.fit_resample(X,Y)

pandas.DataFrame(y_new).value_counts()
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(x_new,y_new,test_size=0.3)
X_train
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

knn = KNeighborsClassifier()
nb = GaussianNB()

model1 = knn.fit(X_train,Y_train)
model2 = nb.fit(X_train,Y_train)

Y_pred1 = model1.predict(X_test)
Y_pred2 = model2.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(Y_test,Y_pred1))
print(accuracy_score(Y_test,Y_pred2))
```
