```python
import numpy
import pandas
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
```

```python
data = pandas.read_csv('twitter_validation.csv',header=None)
df = pandas.DataFrame(data)
df
```

```python
# Add column names
df.columns = ['ID','Entity','Nature of comment','Comment']
df
```

```python
# Convert to lowercase
df['Comment'] = df['Comment'].str.lower()
df
```

```python
import re

text = "python is a language https://www.google.com/"

text = re.sub(r'https\S+',"",text)

text
```

```python
# Remove urls and special characters

def replace_url(text):

    text = re.sub(r'http\S+',"",text)

    text = re.sub(r'[^a-z\s]',"",text)

    return text

df['Comment'] = df['Comment'].apply(replace_url)

df['Comment']
```

```python
# Tokenisation

from nltk.tokenize import word_tokenize

df['Tokens'] = df['Comment'].apply(word_tokenize)

df['Tokens']
```

```python
# Removing stopwords

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

df['Tokens'] = df['Tokens'].apply(lambda a:[i for i in a if i not in stop_words])

df['Tokens']
```

```python
# Lemmitzation

from nltk.stem import PorterStemmer

stemmer = PorterStemmer() # Used for stemming each token in each row

df['Tokens'] = df['Tokens'].apply(lambda a:[stemmer.stem(i) for i in a])

df['Tokens']
```

```python
# Convert back to text
# [mention, facebook, struggl, motiv, go, run, d...] -> "mention facebook struggl motiv go run d..."
df['Text'] = df['Tokens'].apply(lambda a: " ".join(a))
df['Text']
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfid = TfidfVectorizer()

X = tfid.fit_transform(df['Text'])

X
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

Y = le.fit_transform(df['Nature of comment'])

Y
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB

mult = MultinomialNB()

model = mult.fit(X_train,Y_train)

Y_pred = model.predict(X_test)

Y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix

print(accuracy_score(Y_test,Y_pred))

print(confusion_matrix(Y_test,Y_pred))
```
