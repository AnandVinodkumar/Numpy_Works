```python
import numpy
import pandas
```

```python
data = pandas.read_csv('heart.csv')
df = pandas.DataFrame(data)
df
```

```python
df.isna().sum()
```

```python
df = df.dropna(subset=['thal','ca','thalach','trestbps','restecg'])
```

```python
df.isna().sum()
df
```

```python
X = df.iloc[:,:-1]
Y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.3)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)

X_train
X_test
```

```python
from keras.models import Sequential
from keras.layers import Dense

ann = Sequential()

# input layer
ann.add(Dense(units=13,activation='relu'))

# hidden layer 1
ann.add(Dense(units=20,activation='relu'))

# hidden layer 2
ann.add(Dense(units=15,activation='relu'))

# output layer
ann.add(Dense(units=1,activation='sigmoid'))
```

```python
ann.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
ann.fit(X_train,Y_train,epochs=50,batch_size=16)
```

```python
ann.summary()
```

```python
ann.evaluate(X_test,Y_test)
```

```python
Y_pred = ann.predict(X_test)
Y_pred = Y_pred>0.5
Y_pred
```

```python
from sklearn.metrics import classification_report

print(classification_report(Y_test,Y_pred,digits=4))
```

```python
y_new = ann.predict(scaler.transform([[52,1,0,125,212,0,1,168,0,1.0,2,2,3]]))

if y_new > 0.5:
    print("Heart Disease")
else:
    print("No Heart Disease")
```
